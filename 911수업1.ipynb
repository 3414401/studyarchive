````python
#도미
bream_length = [25.4, 26.3, 26.5, 29.0, 29.0, 29.7, 29.7, 30.0, 30.0, 30.7, 31.0, 31.0,
                31.5, 32.0, 32.0, 32.0, 33.0, 33.0, 33.5, 33.5, 34.0, 34.0, 34.5, 35.0,
                35.0, 35.0, 35.0, 36.0, 36.0, 37.0, 38.5, 38.5, 39.5, 41.0, 41.0]
bream_weight = [242.0, 290.0, 340.0, 363.0, 430.0, 450.0, 500.0, 390.0, 450.0, 500.0, 475.0, 500.0,
                500.0, 340.0, 600.0, 600.0, 700.0, 700.0, 610.0, 650.0, 575.0, 685.0, 620.0, 680.0,
                700.0, 725.0, 720.0, 714.0, 850.0, 1000.0, 920.0, 955.0, 925.0, 975.0, 950.0]

import matplotlib.pyplot as plt

plt.scatter(bream_length,bream_weight)
plt.xlabel('length')
plt.ylabel('weight')


#빙어
smelt_length = [ 9.8, 10.5, 10.6, 11.0, 11.2, 11.3, 11.8, 11.8, 12.0, 12.2,
                            12.4, 13.0, 14.3, 15.0]
smelt_weight = [ 6.7, 7.5, 7.0, 9.7, 9.8, 8.7, 10.0, 9.9, 9.8, 12.2, 13.4,
                             12.2, 19.7, 19.9]

import matplotlib.pyplot as plt

plt.scatter(smelt_length,smelt_weight,color='purple') #색지정도 가능!

plt.show()
````

도미,빙어 합치기

````python
#리스트 합치기 (bream, smelt 표기없이 합치지만 순서대로 쌓여서 나중에 zip으로 분류할 수 있음)
length=bream_length+smelt_length
weight=bream_weight+smelt_weight

print(length)
print(weight)

#사이킷런 패키지 사용을 위해! 두 리스트를 2차원 리스트로 만들기
fish_data=[[l,w] for l,w in zip(length, weight)]
print(fish_data)
````


***1. zip()함수***
- 리스트 두개를 튜플로 묶어서 반환. #3개도 되나?

***2. 파이썬의 리스트 컴프리핸션***
-반복문과 조건문을 이용해 **리스트를 쉽고 짧게 한 줄로 생성**하는 문법

ex1.

```
for x in range(1,10):
  numbeerlist.append(x+5)
```
->


```
[x+5 for x in range(1,10)]
```


ex2.(현재 우리 상황)

```
for l,w in zip(length, weight):
  fish_data.append([l,w])
```
->
```
fish_data=[[l,w] for l,w in zip(length, weight)]
```




fit 메서드 - 지도학습이니까 정답 주고 학습시키기

````python
#빙어가 0, 도미가 1
#곱셈연산자를 사용하면 파이썬 리스트를 간단하게 반복시킬 수 있음

#도미 35개, 빙어 0개의 정답 원소만들기
fish_target=[1]*35+[0]*14
````

````python
import sklearn
kn=sklearn.neighbors.KNeighborsClassifier()
````

````python
kn.fit(fish_data,fish_target)
kn.score(fish_data,fish_target)
````

fit메서드 사용...
***- 원리는 나중에 찾아보자!***

- 2차원 배열을 사용한다.
그럼 특징이 4개라면?




```
#훈련 데이터가 3개, 특징이 4개인 경우 (3,4)배열 만듦
X = [
    [25.4, 242, 7.1, 3.2],
    [26.3, 290, 7.5, 3.5],
    [26.5, 340, 8.0, 3.8]
]

kn.fit(X,X_target)
```



특징이 1024개라면:

(100, 1024)

→ 100개 샘플 각각이 1024개의 특징을 가진 것.

fit 메서드
- fit할 때마다 새로 학습한다.


```
kn.fit(train_data, train_target)
kn.score(train_data, train_target)   # 1

kn.fit(test_data, test_target)
kn.score(test_data, test_target)     # 0.7
```

- 오답노트같은거 없다. KNN은 훈련할 때

"이 데이터의 정답은 0이고, 저 데이터의 정답은 1이구나."

라는 **훈련 데이터 자체**를 저장해 둔다.

그리고 새로운 데이터가 들어오면:

**"얘랑 가장 가까운 훈련 데이터가 뭐지?"**

를 찾아서 정답을 결정해.

````python
kn.predict([[30,600]])
kn.predict([[50,600],[40,400]])
#마지막 줄만 결과 나옴. 앞에것도 보고 싶으면 print(kn.predict([[30,600]]))해야함.
````

k=49로 수정한다면? (물고기 데이터 전체가 49개/ 도미 35개 + 빙어 14개)

````python
kn49=sklearn.neighbors.KNeighborsClassifier(n_neighbors=49)

kn49.fit(fish_data,fish_target)
kn49.score(fish_data,fish_target)
````

49개의 점을 모두 참고하므로 score()결과 정확히 35/49인 0.714...가 나온다.
